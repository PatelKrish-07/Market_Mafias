```python
import pandas as pd;
import numpy as np;
import matplotlib.pyplot as plt;
import seaborn as sns;
```

```python
df_clean = pd.read_excel("cleaned_data.xlsx");
```

```python
df_clean
```

```python
daily_revenue = df_clean.groupby(df_clean['InvoiceDate'].dt.date)['TransactionValue'].sum();
daily_revenue
```

```python
monthly_revenue = df_clean.groupby(df_clean['InvoiceDate'].dt.to_period('M'))['TransactionValue'].sum();
monthly_revenue
```

```python
print("\n --Top Products--");
top_products = df_clean.groupby('Description')['Quantity'].sum();
print(top_products.nlargest(5));
```

```python
print("\n --Top customers--");
top_customers = df_clean.groupby('CustomerID')['TransactionValue'].sum().sort_values(ascending=False);
print(top_customers.head(10));
```

```python
print("\n --Country wise revenue--");
country_revenue = df_clean.groupby('Country')['TransactionValue'].sum();
country_revenue
```

```python
print("\n --Average Order Value--");
avg_order_value = df_clean.groupby('InvoiceNo')['TransactionValue'].sum().mean();
print(avg_order_value);
```

```python
print("\n --Number of orders per customer--");
No_of_order_per_customer = df_clean.groupby('CustomerID')['InvoiceNo'].nunique();
print(No_of_order_per_customer);
```

```python
print("\n --Items per Invoice--");
item_per_invoice = df_clean.groupby('InvoiceNo')['Quantity'].sum();
print(item_per_invoice);
```

```python
print("\n --purchase Frequency--");
purchase_freq = df_clean['InvoiceNo'].nunique()/df_clean['CustomerID'].nunique();
print(purchase_freq);
```

```python
print("\n --Customer recency distribution--");
refrence_date = df_clean['InvoiceDate'].max() + pd.Timedelta(days=1);
recency = df_clean.groupby('CustomerID')['InvoiceDate'].max().apply(lambda x: (refrence_date-x).days);
print(recency);
```

```python
print("\n --Cancellation Patterns--");
cancellation_df = pd.read_excel("Cancellations.xlsx");
cancellation_df['YearMonth']=cancellation_df['InvoiceDate'].dt.to_period('M');
monthy_cancellation = cancellation_df.groupby('YearMonth')['InvoiceNo'].nunique();
print(monthy_cancellation);
```
