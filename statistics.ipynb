```python
import pandas as pd;
import numpy as np;
import matplotlib.pyplot as plt;
import seaborn as sns;
from scipy import stats;
```

```python
rfm_df=pd.read_excel("rfm_analysis.xlsx");
customer_features = pd.read_excel("customer_features.xlsx");
```

```python
stats_df = rfm_df[['CustomerID','Monetary','Frequency']].merge(customer_features[['CustomerID','Preffered_Country','Total_Orders','Avg_Order_Value']],
on='CustomerID',
how='left');
```

```python
stats_df
```

```python
#average spending different between customer groups
#checking the customer groups of uk and other countries combined as uk has most number of customers
uk_customers = stats_df[stats_df['Preffered_Country']=='United Kingdom']['Monetary'];
non_uk_customers = stats_df[stats_df['Preffered_Country']!='United Kingdom']['Monetary'];
```

```python
# we use mannwhitneyu test as the transaction dataset has outliers and it is highly skewed
stat1,p_value1=stats.mannwhitneyu(uk_customers,non_uk_customers,alternative='two-sided');
if p_value1<0.05:
    print("H0:UK customers have significantly higher values than non-UK customers");
else:
    print("H1:Geography does not significantly impact total customer spending.");
```

```python
#we use median bcz data is highly skewed
median_orders =stats_df['Total_Orders'].median();
frequent_aov = stats_df[stats_df['Total_Orders']>=median_orders]['Avg_Order_Value'];
infrequent_aov = stats_df[stats_df['Total_Orders']<median_orders]['Avg_Order_Value'];
stat2,p_value2 = stats.mannwhitneyu(frequent_aov,infrequent_aov,alternative='two-sided');
if p_value2<0.05:
    print("H0:Frequent customers have significantly different order values");
else:
    print("H1:customers have No significantly different order values");
```

```python
#Is purchase frequency associated with monetary value?
corr_coef,p_value3 = stats.spearmanr(stats_df['Frequency'],stats_df['Monetary']);
if p_value3<0.05:
    print("More the frequency, more is the Moneary value");
else:
    print("No correlation of frequency and Monetary value as such");
```
