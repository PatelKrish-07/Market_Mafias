```python
import pandas as pd;
import numpy as np;
import matplotlib.pyplot as plt;
import seaborn as sns;
```

```python
df = pd.read_excel("Updates_transaction.xlsx")
```

```python
print("\n--No of missing Customers--");
print(df['CustomerID'].isna().sum());
```

```python
df_clean = df.dropna(subset=['CustomerID']);
print(f"\n Shape of dataset After removing missing CustomerId's {df_clean.shape}");
```

```python
print("\n --Duplicates--");
print(df_clean.duplicated().sum());
```

```python
print("\n Dropping all the duplicates");
duplicate_df = df_clean[df_clean.duplicated(keep=False)];
df_clean=df_clean.drop_duplicates();
print(df_clean.duplicated().sum());
```

```python
df_clean.shape
```

```python
print("\n--Negative price values and quantities--");
print(df_clean[df_clean['Quantity']<0]['Quantity'].count());
print(df_clean[df_clean['TransactionValue']<0]['TransactionValue'].count());

```

```python
cancellations_df= df_clean[df_clean['InvoiceNo'].str.startswith('C',na=False)].copy()
cancellations_df.shape
```

```python
cancellations_df
```

```python
df_clean = df_clean[~df_clean['InvoiceNo'].astype(str).str.startswith('C')]
```

```python
df_clean.shape
```

```python
print(df_clean[df_clean['Quantity']<0]['Quantity'].count());
print(df_clean[df_clean['TransactionValue']<0]['TransactionValue'].count());
```

```python
print("\n --Removing items with 0 as transaction--");
df_clean=df_clean[(df_clean['Quantity']>0) & (df_clean['UnitPrice']>0)]
print(df_clean.shape);
df_clean
```

```python
print("\n --Standardize Description--");
df_clean['Description']=df_clean['Description'].str.strip().str.upper();
df_clean['Description']
```
